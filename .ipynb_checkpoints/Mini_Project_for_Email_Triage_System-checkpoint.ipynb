```python
import ollama
import random
import json
import re
import time
```

```python
issue_topics = [
    ("fan_not_working", "the fan in my room is not working"),
    ("wifi_down", "the wifi is down or very slow"),
    ("water_leakage", "there is water leakage in my room or bathroom"),
    ("electricity_issue", "the electricity is tripping frequently"),
    ("mess_food", "the mess food quality is bad"),
    ("ac_not_cooling", "the AC is not cooling properly"),
    ("broken_window", "the window latch is broken"),
    ("dirty_washroom", "the washroom is not getting cleaned"),
    ("billing_issue", "there is a fee or billing confusion"),
    ("noise_complaint", "the room next door is very noisy"),
]

issue_labels = [x[0] for x in issue_topics]

ISSUE_KEYWORDS = {
    "fan_not_working": ["fan"],
    "wifi_down": ["wifi", "internet"],
    "water_leakage": ["water", "leak"],
    "electricity_issue": ["electric", "power"],
    "mess_food": ["mess", "food"],
    "ac_not_cooling": ["ac", "cool"],
    "broken_window": ["window"],
    "dirty_washroom": ["washroom", "dirty", "bathroom"],
    "billing_issue": ["bill", "fee"],
    "noise_complaint": ["noise", "loud"],
}

URGENCY_REQUIRED = {
    "low": ["not urgent", "no rush", "whenever possible"],
    "medium": ["soon", "next few days", "this week"],
    "high": ["urgent", "as soon as possible", "immediate", "before"]
}

day_phrases = [
    "before tomorrow morning",
    "before Monday",
    "before this weekend",
    "within today if possible"
]

subject_templates = [
    "Issue in Room {room_no}",
    "Maintenance Required in Room {room_no}",
    "Request for Assistance - Room {room_no}",
    "Room {room_no} Facility Problem",
    "Hostel Maintenance Request for Room {room_no}"
]
```

```python
def call_model(model, prompt, timeout=15, max_retries=2):
    for attempt in range(max_retries):
        start = time.time()
        try:
            response = ollama.chat(
                model=model,
                messages=[{"role": "user", "content": prompt}],
                stream=False
            )
            return response["message"]["content"]
        except:
            pass

        if time.time() - start > timeout:
            print(f"Timeout on {model}, retry {attempt+1}/{max_retries}")

    # If all retries fail, return empty string
    return ""
```

```python
def generate_raw_body(labels, urgency_phrase):
    issues = ", ".join(
        desc for (label, desc) in issue_topics if label in labels
    )

    prompt = f"""
Write 3 short sentences describing these hostel issues:
{issues}

Rules:
- Present tense
- Simple student language
- No greetings
- No thanks
- No subject
- No extra explanations
- Include urgency phrase: "{urgency_phrase}"

Write only the sentences.
"""

    return call_model("qwen2.5:3b", prompt, timeout=12)

```

```python
def rewrite_body_clean(raw):
    prompt = f"""
Rewrite the following into a clean, semi-formal, present-tense hostel complaint email body.

Rules:
- 4–5 sentences
- No greeting
- No closing or thanks
- No signature
- No subject line
- No repeated sentences
- No placeholders
- No markdown
- Clean simple English

Raw text:
"{raw}"

Write only the body text.
"""

    return call_model("llama3.2:3b", prompt)
```

```python
def clean_body(text):
    # Remove forbidden sections
    text = re.sub(r"(Hi|Hello|Dear)[^\n]*", "", text, flags=re.IGNORECASE)
    text = re.sub(r"(Thanks|Regards|Sincerely)[\s\S]*", "", text, flags=re.IGNORECASE)
    text = re.sub(r"\[[^\]]*\]", "", text)
    text = re.sub(r"\{[^\}]*\}", "", text)

    # Normalize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # Split sentences safely
    sentences = re.split(r"(?<=[.!?])\s+", text)
    sentences = [s.strip() for s in sentences if s.strip()]

    # Keep max 5 complete sentences
    sentences = sentences[:5]

    body = " ".join(sentences)
    return body.strip()

def is_valid_body(body, labels, urgency):
    lower = body.lower()

    # Sentence completeness
    if not sentences_are_complete(body):
        return False

    # Sentence count
    sentences = re.split(r"(?<=[.!?])\s+", body)
    if len(sentences) < 3 or len(sentences) > 5:
        return False

    # Sentence length sanity
    for s in sentences:
        if len(s.split()) < 5:
            return False

    # Issue keyword enforcement
    for label in labels:
        if not any(kw in lower for kw in ISSUE_KEYWORDS[label]):
            return False

    # Block fake causal logic between issues
    forbidden_links = ["because", "causing", "due to", "therefore", "as a result"]
    for fl in forbidden_links:
        if fl in lower:
            return False

    # Urgency consistency
    if not any(u in lower for u in URGENCY_REQUIRED[urgency]):
        return False

    return True

def sentences_are_complete(body):
    # Must end with proper punctuation
    if not body.endswith((".", "!", "?")):
        return False

    # Reject dangling fragments
    bad_endings = (" t", " w", " th", " re", " ca", " du")
    for be in bad_endings:
        if body.lower().endswith(be):
            return False

    return True
```

```python
def build_email(body, room_no):
    subject = random.choice(subject_templates).format(room_no=room_no)
    recipient = random.choice(["Sir", "Madam", "Warden", "Maintenance Team"])
    sender = random.choice(["Rohan", "Amit", "Sneha", "Priya", "Karan", "Nisha", "Rahul"])

    return f"Subject: {subject}\n\nHi {recipient},\n\n{body}\n\nThanks,\n{sender}\nRoom {room_no}"
```

```python
TOTAL = 50
dataset = []

for i in range(TOTAL):
    labels = random.sample(issue_labels, random.choice([1, 2]))
    urgency = random.choice(["low", "medium", "high"])

    if urgency == "high":
        urgency_phrase = random.choice(URGENCY_REQUIRED["high"] + day_phrases)
    else:
        urgency_phrase = random.choice(URGENCY_REQUIRED[urgency])

    attempts = 0
    body = None

    while attempts < 3:
        raw = generate_raw_body(labels, urgency_phrase)

        if not raw.strip():
            print("Skipping due to empty raw body")
            attempts += 1
            continue

        rewritten = rewrite_body_clean(raw)
        body = clean_body(rewritten)

        if is_valid_body(body, labels, urgency):
            break

        attempts += 1

    if body is None or not is_valid_body(body, labels, urgency):
        print("Skipping one…")
        continue

    room_no = random.randint(101, 520)
    email = build_email(body, room_no)

    dataset.append({
        "email_text": email,
        "labels": labels,
        "urgency": urgency
    })

    print(f"Generated {i+1}/{TOTAL}")
```

```python
with open("hostel_dataset_v5_50.json", "w", encoding="utf-8") as f:
    json.dump(dataset, f, indent=2)

print("SAVED: hostel_dataset_v5_50.json")
```
